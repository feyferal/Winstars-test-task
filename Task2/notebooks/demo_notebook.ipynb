# Multimodal Animal Recognition Pipeline

This task demonstrates a multimodal pipeline that combines:
- Named Entity Recognition (NER) for extracting animal names from text
- Computer Vision (CV) for classifying images

The goal is to determine whether the text and image refer to the same animal.

## Approach

Pipeline:

Text → NER → animal
Image → CV → animal
→ Compare results → match / mismatch
